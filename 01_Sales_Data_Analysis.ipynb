```python
## Sequencing of month
```

```python
import calendar

monthly_sales = df.groupby('Order Month')['Sales'].sum()

monthly_sales.index = monthly_sales.index.map(
    lambda x: calendar.month_name[x]
)

monthly_sales
```

```python
## Create a line with the mmonthly_sales
```

```python
plt.figure(figsize=(12, 5))

plt.plot(monthly_sales.index, monthly_sales.values, marker = 'o')

plt.title ('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales ($)')

plt.xticks(rotation = 45)
plt.grid(True)

plt.show()
```

```python
## for ccontainer in ax.containers:  :- Sales aur profit ke dono bar group ko ek ek kerke access kerta hai
## bar_label():- bar ke uper values show karega, Padding= 3 ka matlab bar aur value ke bich space
```

```python
category_analysis = df.groupby('Category').agg(
    Total_Sales = ('Sales', 'sum'),
    Total_Profit = ('Profit', 'sum')
)
ax = category_analysis.plot(kind = 'bar', figsize = (9 ,5))

plt.title('Category-wise Sales vs Profit')

plt.xlabel('Category')
plt.ylabel('Amount ($)')
plt.xticks(rotation=0)

for container in ax.containers: ax.bar_label(container, fmt= '%.0f', padding = 3)

plt.tight_layout()
plt.show()
```

```python
## Customer-wise Average Order Value (AOV)
```

```python
customer_aov = df.groupby('Customer Name').agg(
    Total_Sales = ('Sales', 'sum'),
    Total_Orders = ('Order ID', 'nunique')
)

customer_aov['Average_Order_Value'] = (customer_aov['Total_Sales'] / customer_aov['Total_Orders'])

customer_aov.sort_values('Average_Order_Value', ascending = False).head(10)
```

```python
top_10_customer_aov = customer_aov.sort_values('Average_Order_Value', ascending= False).head(10)

ax = top_10_customer_aov.plot(
    kind = 'bar', figsize = (12, 7)
)

plt.title('Top 10 Customer by AOV')
plt.xlabel('Customer Name')
plt.ylabel('Total Sales, Total Order & AOV')
plt.xticks(rotation = 45, ha = 'right')

for container in ax.containers: ax.bar_label(container, fmt='%.0f', padding=3)
plt.legend(title = 'Business Metrics')

plt.tight_layout()
plt.show()
```
